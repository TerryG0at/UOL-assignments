## 1. Introduction to Domain-specific area

#### Sentiment Analysis in Movie Reviews

Sentiment analysis is one of the most important and widely used tasks in Natural Language Processing (NLP). In simple terms, it refers to the process of automatically analysing text to determine whether the expressed opinion is positive, negative, or neutral (Liu, 2012). This type of analysis is especially valuable in the movie industry. With global box office revenues exceeding $100 billion each year and the rapid growth of streaming platforms, understanding how audiences feel about films has become increasingly important for studios, distributors and content creators. Movie reviews are well suited for sentiment analysis because people who gave reviews usually express clear opinions about specific films which make the overall sentiment relatively easy to identify (Pang and Lee, 2008).

Over the past two decades sentiment analysis methods have been developed significantly. Earlier methods mainly relied on traditional machine learning algorithms such as Naive Bayes and Support Vector Machines. These approaches treated text as a collection of individual words and counted how often each word appeared. This method is often known as the bag-of-words model. While these techniques produced reasonable results, they often failed to capture deeper meaning, context or subtle language features such as sarcasm (Pang et al., 2002). Later technique introduced word embedding methods such as Word2Vec and GloVe, which represent words as numerical vectors that reflect their meanings and relationships to other words. This allowed models to better understand context and similarity between words rather than relying only on word frequency (Mikolov et al., 2013). Despite these improvements, researchers continue to debate whether newer embedding-based approaches are always superior to traditional methods, especially when considering factors such as computational cost, simplicity and interpretation.

Movie review sentiment analysis holds strong academic significance. The seminal work by Pang, Lee and Vaithyanathan (2002) demonstrated that identifying sentiment is fundamentally different from simply identifying a document’s topic. Their work showed that machine learning could successfully classify opinions in movie reviews, achieving an accuracy of around 83% using Naive Bayes. This study established widely used benchmarks that continue to influence research today. More recent approaches that use word embeddings and advanced models have reported accuracy levels above 90%, although these improvements often come with higher computational demands (Maas et al., 2011). As a result, movie reviews remain an excellent case study for comparing traditional and modern sentiment analysis methods, offering clear performance benchmarks and a rich body of existing research.

### 5.2 Text Preprocessing

```python
# Initialize NLTK tools
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_text_statistical(text):
    
    # Convert to lowercase
    text = str(text).lower()
    
    # Remove HTML tags
    text = re.sub(r'<[^>]+>', '', text)
    
    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)
    
    # Remove special characters and digits
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    
    # Tokenization
    tokens = word_tokenize(text)
    
    # Remove stop words, lemmatize and filter short words
    processed_tokens = [
        lemmatizer.lemmatize(word) 
        for word in tokens 
        if word not in stop_words and len(word) > 2
    ]
    
    # Join back into string
    return ' '.join(processed_tokens)

def preprocess_text_embedding(text):

    # Convert to lowercase
    text = str(text).lower()
    
    # Remove HTML tags
    text = re.sub(r'<[^>]+>', '', text)
    
    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)
    
    # Keep alphanumeric and spaces
    # This preserves numbers and context
    text = re.sub(r'[^\w\s]', ' ', text)
    
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

# Apply statistical preprocessing
df['review_statistical'] = df['review'].apply(preprocess_text_statistical)
print("Statistical preprocessing complete")

# Apply embedding preprocessing
df['review_embedding'] = df['review'].apply(preprocess_text_embedding)
print("Embedding preprocessing complete")

print("\nShowing 3 random examples to illustrate the differences:")

for example_num in range(3):
    sample_idx = np.random.randint(0, len(df))
    original = df['review'].iloc[sample_idx]
    stat_result = df['review_statistical'].iloc[sample_idx]
    emb_result = df['review_embedding'].iloc[sample_idx]
    
    print(f"\n(Review #{sample_idx})")
    
    print(f"\nOriginal Review ({len(original)} chars, {len(original.split())} words):")
    print(f"{original[:200]}...")
    
    print(f"\nStatistical Preprocessing ({len(stat_result)} chars, {len(stat_result.split())} words):")
    print(f"{stat_result[:200]}")
    
    print(f"\nEmbedding Preprocessing ({len(emb_result)} chars, {len(emb_result.split())} words):")
    print(f"{emb_result[:200]}")

    print("="*80)
```

- Lowercase conversion ensures that words such as “Movie” and “movie” are treated as the same token which reduce unnecessary duplication in the vocabulary.

- HTML tag removal eliminates markup elements like <br> <p> that appear in some reviews but do not contribute to the analysis.

- URL removal prevents web links from being taken as tokens as they add noise and do not contribute in frequency-based models.

- Special character and digit removal keeps only alphabetic characters, simplifying the vocabulary and preventing punctuation or numbers from introducing irrelevant features.

- Whitespace normalization ensures consistent spacing between words and prevent accidental creation of distinct tokens.

- Tokenization splits the cleaned text into individual words enabling filtering and feature extraction required by TF-IDF models.

- Stop word removal and lemmatization reduce vocabulary size and improve feature quality by removing common words like "a" "the" and converting words to their base form (example: “loved” → “love”).

This project implements Two different preprocessing approaches because different models have different requirements.

1. Statistical Preprocessing (for TF-IDF models):
- Heavy cleaning to reduce vocabulary and noise
- Stop word removal, lemmatization, aggressive filtering
- Creates clean, reduced text for frequency counting
- Used by: Baseline (Logistic Regression) and Naive Bayes
   
2. Embedding Preprocessing (for CNN model):
- Minimal cleaning to preserve context
- Keeps stop words, no lemmatization, light filtering
- Preserves richer text for semantic understanding
- Used by: CNN with pre-trained GloVe embeddings

Why two approaches?

Statistical models (TF-IDF):
- Count word frequencies
- Need aggressive cleaning to reduce dimensionality
- Stop words ("the", "a") don't help frequency-based classification
  
Deep learning models (CNN):
- Learn patterns from context
- Pre-trained embeddings already handle word variations
- Stop words provide important contextual information
- CNN filters can learn "not good" as a negative pattern

```python
# Load GloVe Embeddings (for CNN)

def load_glove_embeddings(filepath):
    embeddings = {}
    
    print(f"Loading GloVe embeddings from: {filepath}")
    
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            for line in f:
                values = line.split()
                word = values[0]
                vector = np.asarray(values[1:], dtype='float32')
                embeddings[word] = vector
        
        print(f"Loaded {len(embeddings):,} word embeddings")
        return embeddings
        
    except FileNotFoundError:
        print(f"\nError: {filepath} not found!")
        raise

# Load GloVe
embeddings = load_glove_embeddings('glove.6B.100d.txt')
embedding_dim = len(next(iter(embeddings.values())))

print(f"\nGloVe embeddings loaded:")
print(f"Vocabulary: {len(embeddings):,} words")
print(f"Dimension: {embedding_dim}d")

# Show example embeddings
print("\nExample word vectors:")
for word in ['great', 'terrible', 'movie', 'film']:
    if word in embeddings:
        vec = embeddings[word]
        print(f"'{word}': [{vec[0]:.3f}, {vec[1]:.3f}, {vec[2]:.3f}, ..., {vec[-1]:.3f}]")
```

## 5.3 Text Representation

For statistical model, i will use TF-IDF. It converts text into numerical vectors by measuring word importance. Term Frequency (TF) measures how often a word appear in a review. Formula for TF score: TF = total appearance of the word / total words in review. Example, if "bad" appears 2 times in 20 words review, the TF(bad) is 2/20 = 0.1. Inverse Document Frequency (IDF) measures the frequency of the word across all reviews. Formula for IDF score: IDF = log(total reviews / number of reviews containing that word). Example, if "bad" appears in 10,000 out of 50,000 reviews the IDF score is log(50,000/10,000) = 0.69. TF-IDF score is calculated as follow: TF-IDF = TF x IDF. High TF-IDF scores usually mean the word is frequent in this review but rare across all the reviews while Low scores describe that the word is either rare in this review or too common everywhere like the stop words which appear almost everywhere.

Example review: "This movie was great! Great acting, great plot."
- "great": appears 3 times, somewhat rare -> High TF-IDF
- "movie": appears 1 time, very common -> Medium TF-IDF
- "bad": appears 0 times -> Zero TF-IDF
- "was": appears 1 time, very common -> Low TF-IDF

I use TF-IDF as it highlights some distinctive words essential for classification, reduce the weights of common words like "a" "and" "of" and creates sparse numerical vectors suitable for ML models.

## 5.4 Train-Test split

```python
# Create features and labels
# Heavy preprocessing (for TF-IDF models)
X_statistical = df['review_statistical'].values
# Light preprocessing (for embedding model)
X_embedding = df['review_embedding'].values
y = df['sentiment'].values

print(f"Total dataset: {len(df):,} reviews")
print(f"Sentiment distribution:")
print(pd.Series(y).value_counts())

# Stratified split (80-20) for statistical preprocessing
X_train, X_test, y_train, y_test = train_test_split(
    X_statistical, y,
    # 20% for testing
    test_size=0.2,
    # For reproducibility
    random_state=42,
    # Maintain 50-50 balance
    stratify=y
)

print(f"\nStatistical preprocessing:")
print(f"Training set: {len(X_train):,} reviews")
print(f"Test set:     {len(X_test):,} reviews")

# Same split for embedding preprocessing
X_train_emb, X_test_emb, _, _ = train_test_split(
    X_embedding, y,
    test_size=0.2,
    # Same random state = same split
    random_state=42,
    stratify=y
)

print(f"\nEmbedding preprocessing:")
print(f"Training set: {len(X_train_emb):,} reviews")
print(f"Test set:     {len(X_test_emb):,} reviews")

print(f"\nTraining set sentiment distribution:")
print(pd.Series(y_train).value_counts())

print(f"\nTest set sentiment distribution:")
print(pd.Series(y_test).value_counts())
```

The dataset is split into a training set and a test set to separate model learning from performance evaluation. The training set is used to teach the model patterns in the data while the test set is used to assess how well the model perform to unseen reviews. Evaluating the model on the same data used for training would lead to high accuracy and this does not reflect real world performance.

An 80–20 split is used because allocating 80% of the data to training provides sufficient examples for the model to learn meaningful patterns. The remaining 20% provides large enough test set to produce reliable and stable evaluation metrics. This split ratio is widely used in machine learning practice.

A stratified is applied to ensure that both the training and test sets maintain the original 50–50 balance between positive and negative reviews. Without stratification, a purely random split could result in class imbalance such as a training set containing a higher proportion of positive reviews than the test set. Stratification guarantees that model evaluation remains fair and that performance metrics are not biased by uneven class distributions.

A fixed random_state value of 42 is used to make the data split reproducible. This ensures that running the code multiple times produces the same training and test sets.

```python
# CNN Data Preparation after Train-Test split

# Initialize tokenizer
max_words = 20000
max_length = 200

tokenizer = Tokenizer(num_words=max_words, oov_token='<OOV>')

print("Fitting tokenizer on training set")
tokenizer.fit_on_texts(X_train_emb)

print()
print(f"Vocabulary built from training data")
print(f"Total unique words found: {len(tokenizer.word_index):,}")
print(f"Using top {max_words:,} most frequent words")

X_train_seq = tokenizer.texts_to_sequences(X_train_emb)
X_test_seq = tokenizer.texts_to_sequences(X_test_emb)

print(f"Text converted to integer sequences")
print(f"Training sequences: {len(X_train_seq):,}")
print(f"Test sequences: {len(X_test_seq):,}")

# Pad Sequences

# Check sequence length distribution
seq_lengths = [len(seq) for seq in X_train_seq]
print(f"\nSequence length distribution (before padding):")
print(f"Mean:   {np.mean(seq_lengths):.1f} words")
print(f"Median: {np.median(seq_lengths):.1f} words")
print(f"Max:    {max(seq_lengths)} words")
print(f"Min:    {min(seq_lengths)} words")

truncated_count = sum(1 for l in seq_lengths if l > max_length)
print(f"Sequences > {max_length}: {truncated_count:,} ({truncated_count/len(seq_lengths)*100:.1f}% will be truncated)")

print(f"\nPadding all sequences to {max_length} words")
X_train_padded = pad_sequences(X_train_seq, maxlen=max_length, padding='post', truncating='post')
X_test_padded = pad_sequences(X_test_seq, maxlen=max_length, padding='post', truncating='post')

print(f"Sequences padded to fixed length")
print(f"Training shape: {X_train_padded.shape} ({X_train_padded.shape[0]:,} samples × {X_train_padded.shape[1]} words)")
print(f"Test shape:     {X_test_padded.shape} ({X_test_padded.shape[0]:,} samples × {X_test_padded.shape[1]} words)")

# Convert Labels to Binary
print("Converting string labels to binary")
y_train_binary = (y_train == 'positive').astype(int)
y_test_binary = (y_test == 'positive').astype(int)

print(f"Labels converted to binary")
print(f"Mapping: 'positive' -> 1, 'negative' -> 0")

print(f"\nLabel distribution:")
print(f"Training set:")
print(f"Positive (1): {sum(y_train_binary):,} ({sum(y_train_binary)/len(y_train_binary)*100:.1f}%)")
print(f"Negative (0): {len(y_train_binary) - sum(y_train_binary):,} ({(len(y_train_binary) - sum(y_train_binary))/len(y_train_binary)*100:.1f}%)")
print(f"Test set:")
print(f"Positive (1): {sum(y_test_binary):,} ({sum(y_test_binary)/len(y_test_binary)*100:.1f}%)")
print(f"Negative (0): {len(y_test_binary) - sum(y_test_binary):,} ({(len(y_test_binary) - sum(y_test_binary))/len(y_test_binary)*100:.1f}%)")

# Create Embedding Matrix
# Get vocabulary info from tokenizer
word_index = tokenizer.word_index
vocab_size = min(max_words, len(word_index)) + 1  # +1 for padding token (index 0)
print(f"Matrix shape: ({vocab_size:,} words, {embedding_dim} dimensions)")

# Initialize matrix with zeros
embedding_matrix = np.zeros((vocab_size, embedding_dim))

# Fill matrix with GloVe vectors
found_count = 0
for word, i in word_index.items():
    # Skip words beyond the vocabulary size
    if i >= max_words:
        continue
    
    # Look up word in GloVe embeddings
    embedding_vector = embeddings.get(word)
    if embedding_vector is not None:
        embedding_matrix[i] = embedding_vector
        found_count += 1

print(f"Embedding matrix created")
print(f"Total vocabulary size: {vocab_size:,} words")
print(f"Words with GloVe vectors: {found_count:,} ({found_count/vocab_size*100:.1f}%)")
print(f"Words without GloVe (zeros): {vocab_size - found_count:,} ({(vocab_size - found_count)/vocab_size*100:.1f}%)")
```

##### Baseline Model: Logistic Regression with TF-IDF Features

The baseline model combines Logistic Regression with Term Frequency-Inverse 
Document Frequency (TF-IDF) feature representation (Salton & Buckley, 1988). 
This approach offers several advantages:

- easy to implement and understand
- fast training
- require minimal hyperparameter tuning
- achieve acceptable performance
- Used in Pang et al. (2002) foundational work
- Industry standard for text classification baselines
- More sophisticated than simple word counting
- Less complex than neural networks

TF-IDF weighting captures the importance of words within documents while accounting for their frequency across the entire corpus (Salton & Buckley, 1988). Words that appear frequently in a specific document but rarely across all documents receive higher weights, making them strong discriminative features for classification. Combined with Logistic Regression, this provides a robust baseline against which more complex models can be evaluated (Pang et al., 2002).

## 6.2 TF-IDF Features

```python
# Initialize TF-IDF vectorizer
vectorizer = TfidfVectorizer(
    # Keep top 5000 most important features
    max_features=5000,
    # Use both single words and bigrams
    ngram_range=(1, 2),
    # Ignore words appearing in less than 2 reviews
    min_df=2,
    # Ignore words appearing in more than 95% of reviews
    max_df=0.95,
    # Apply log scaling to term frequency
    sublinear_tf=True
)

# Fit vectorizer on training data and transform both sets
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print(f"Feature matrix shape:")
print(f"Training: {X_train_tfidf.shape}")
print(f"Testing:  {X_test_tfidf.shape}")
print(f"Sparsity: {(1.0 - X_train_tfidf.nnz / (X_train_tfidf.shape[0] * X_train_tfidf.shape[1])) * 100:.1f}%")
```

Accuracy is calculated as the proportion of correct predictions among all predictions made by the model. It provides an overall measure of classification correctness and is particularly appropriate for balanced datasets. For example, an accuracy of 89% means that the model correctly classifies 89% of all reviews.

Precision measures the proportion of reviews predicted as positive that are actually positive. It reflects how reliable positive predictions are and is especially important in scenarios where false positives carry significant costs. For instance, a precision score of 88% indicates that when the model predicts a positive review, it is correct 88% of the time.

Recall measures the proportion of actual positive reviews that are correctly identified by the model. It captures the model’s ability to find all relevant positive instances and is important when false negatives are costly. For example, a recall of 86% means that the model successfully identifies 86% of all positive reviews.

F1-score is calculated as the harmonic mean of precision and recall, then combine both metrics into a single value. It provides a balanced assessment of model performance by penalising extreme imbalances between precision and recall. As a result, the F1-score serves as a useful single-number summary when both false positives and false negatives are important.

## 7.1 Naive Bayes (Statistical Model)

Naive Bayes is a probabilistic classifier based on Bayes' theorem with a "naive" assumption that all features (words) are independent of each other given the class label (McCallum & Nigam, 1998). Despite this simplifying assumption, Naive Bayes has proven remarkably effective for text classification tasks.

Bayes' theorem:
P(sentiment | review) = (P(review | sentiment) × P(sentiment)) / P(review)

For text classification, the Multinomial Naive Bayes variant treats documents as collections of word counts, making it particularly suitable for sentiment analysis (McCallum & Nigam, 1998).

##### Strength
- Fast training
- Fast inference
- Works well with small datasets
- Probabilistic outputs
- Theoretically grounded (Bayes' theorem)
- Handles high-dimensional sparse features well

##### Weakness
- Naive independence assumption rarely true
- Can't capture word interactions ("not good" vs "good")
- Zero probabilities for unseen words
- Generally lower accuracy than discriminative models
- Sensitive to feature quality

#### Reference:

McCallum, A., & Nigam, K. (1998). A Comparison of Event Models for Naive Bayes Text Classification. In AAAI-98 Workshop on Learning for Text Categorization, pages 41–48.

## 7.1.1 Naive Bayes Training 

```python
from sklearn.naive_bayes import MultinomialNB
import time

# Reuse the TF-IDF features from baseline
print(f"Training set: {X_train_tfidf.shape}")
print(f"Test set: {X_test_tfidf.shape}")

# Initialize Multinomial Naive Bayes
nb_model = MultinomialNB(
    # Laplace smoothing (avoids zero probabilities)
    alpha=1.0,
    # Learn class prior probabilities from data
    fit_prior=True
)

# Train the model
print("\nTraining Naive Bayes")
start_time = time.time()
nb_model.fit(X_train_tfidf, y_train)
nb_training_time = time.time() - start_time

print(f"Training complete in {nb_training_time:.4f} seconds")
# Compare training time with baseline
time_diff = training_time - nb_training_time

if nb_training_time < training_time:
    print(f"Naive Bayes is faster than the baseline by {time_diff:.4f} seconds")
elif nb_training_time > training_time:
    print(f"Naive Bayes is slower than the baseline by {abs(time_diff):.4f} seconds")
else:
    print("Naive Bayes training time is equal to the baseline")
```

In Multinomial Naive Bayes, predictions are based on the multiplication of conditional probabilities for each word in a review. A problem arises when a word appears in reviews of one sentiment class but never appears in the other. For example, if the word “excellent” appears frequently in positive reviews but never appears in negative reviews, the estimated probability becomes zero. When a review contains this word, the probability of the negative class is multiplied by zero and cause the entire class probability to fall to zero regardless of other words in the review.

Laplace smoothing (also called additive smoothing) addresses this issue by adding a small constant (typically 1) to every word count before probability estimation (Manning & Schütze, 1999).

#### Example

Without smoothing: P(excellent | negative) = 0 / total_words_in_negative_reviews

With Laplace smoothing: P(excellent | negative) = (0 + 1) / (total_words_in_negative_reviews + vocabulary_size)

This adjustment ensures no probability is exactly zero while maintaining appropriately low probabilities for words genuinely rare in a class. The denominator increases by the vocabulary size to maintain proper probability normalization (Manning & Schütze, 1999).

#### Reference:

Manning, C. D., & Schütze, H. (1999). Foundations of Statistical Natural Language Processing. MIT Press, Cambridge, MA.

Investigation revealed this stems from the interaction between preprocessing and Naive Bayes' probabilistic framework. Our preprocessing pipeline removes stop words, reducing average review length from approximately 200 to 110 words. While this improves feature quality for discriminative models like Logistic Regression (89.46% accuracy), it reduces confidence for generative models like Naive Bayes.

Mathematical Explanation:
Naive Bayes calculates P(sentiment | review) by multiplying individual word probabilities:
P(pos|review) = P(w1|pos) × P(w2|pos) × ... × P(w(n)|pos)
With fewer words there are fewer numbers to work around. Stop words although doesn't contribute much to prediction still contribute to probability mass. Their removal leaves only highly discriminative words (e.g., 'excellent', 'terrible') which the model weights heavily but with less cumulative evidence to support strong confidence.

Implications:
- The model makes correct predictions (86% accuracy) but expresses uncertainty
- This demonstrates a preprocessing trade-off: optimizing for one model type may not optimize for another
- In production, these low-confidence predictions would benefit from human review, revealing 18.4% of classifications as genuinely ambiguous
- This transparency about uncertainty is a unique advantage of probabilistic models

## 7.2 CNN with Word Embeddings

Convolutional Neural Networks (CNNs), originally designed for image processing, 
have proven highly effective for text classification tasks. Unlike traditional 
statistical models that treat text as independent word frequencies, CNNs learn 
local patterns (n-grams) through convolutional filters while leveraging 
pre-trained word embeddings for semantic understanding.

Embeddings: Dense vector representations where semantically similar words have similar vectors.

##### Example semantic relationships
- king - man + woman ~ queen
- Paris - France + Germany ~ Berlin

##### In our model
- Pre-trained GloVe vectors (400,000 words, 100 dimensions) (Pennington et al., 2014)
- Fine-tuned during CNN training for sentiment task
- Transfer learning: leverage knowledge from 6B token corpus (Pennington et al., 2014)

##### Strengths:

- N-gram Pattern Recognition: Captures multi-word sentiment phrases
- Semantic Understanding: Embeddings recognize synonyms and related words
- Position Invariance: Max pooling detects important features anywhere in text
- Transfer Learning: Pre-trained embeddings provide semantic knowledge
- Computational Efficiency: Faster than sequential models (LSTM/RNN)
- Dense Features: 100-d embeddings vs 5000-d sparse TF-IDF

##### Weaknesses:

- Requires Pre-trained Embeddings: Dependency on external resource (GloVe)
- Out-of-Vocabulary Words: Unknown words get zero vectors
- Training Time: 7 minutes vs <1 second for TF-IDF
- Less Interpretable: Neural network "black box" vs explicit TF-IDF weights
- Fixed Context Window: Filters limited to 3-5 word phrases
- Memory Requirements: Embedding matrix + model parameters

#### Why This Architecture (Kim 2014)?

The multi-filter CNN architecture follows Kim (2014)'s influential paper "Convolutional Neural Networks for Sentence Classification" which demonstrated that simple CNNs with multiple filter sizes achieve strong performance on sentiment analysis tasks.

##### Key innovations
1. Multiple filter sizes (3, 4, 5) capture different n-gram patterns
2. Global max pooling preserves strongest signal regardless of position
3. Pre-trained embeddings provide semantic foundation
4. Simple architecture trains quickly while remaining effective

#### References:

Kim, Y. (2014). Convolutional Neural Networks for Sentence Classification. 
In Proceedings of the 2014 Conference on Empirical Methods in Natural Language 
Processing (EMNLP), pages 1746–1751, Doha, Qatar. Association for Computational 
Linguistics.

Pennington, J., Socher, R., & Manning, C. D. (2014). GloVe: Global Vectors for 
Word Representation. In Proceedings of the 2014 Conference on Empirical Methods 
in Natural Language Processing (EMNLP), pages 1532–1543, Doha, Qatar. Association 
for Computational Linguistics.

## 7.2.1 CNN Model Training

```python
# Input layer
input_layer = Input(shape=(max_length,), name='input')

# Embedding layer
embedding_layer = Embedding(
    input_dim = vocab_size,
    output_dim = embedding_dim,
    weights = [embedding_matrix],
    input_length = max_length,
    trainable = True,
    name = 'embedding'
)(input_layer)

# Spatial dropout
dropout_emb = SpatialDropout1D(0.2)(embedding_layer)

# Multiple CNN filters
filter_sizes = [3, 4, 5]
num_filters = 128
conv_blocks = []

for filter_size in filter_sizes:
    conv = Conv1D(filters = num_filters, kernel_size = filter_size, activation = 'relu', padding = 'valid')(dropout_emb)
    pool = GlobalMaxPooling1D()(conv)
    conv_blocks.append(pool)

# Concatenate
concatenated = Concatenate()(conv_blocks)

# Dense layers
dropout = Dropout(0.5)(concatenated)
dense = Dense(128, activation = 'relu')(dropout)
output = Dense(1, activation = 'sigmoid')(dense)

# Create and compile model
cnn_model = Model(inputs = input_layer, outputs = output, name = 'CNN_Sentiment')
cnn_model.compile(optimizer = Adam(learning_rate = 0.001), loss = 'binary_crossentropy', metrics = ['accuracy'])

print("Model architecture built")
print(f"Parameters: {cnn_model.count_params():,}")
print("\nModel Summary:")
cnn_model.summary()

# Callbacks
callbacks_cnn = [
    EarlyStopping(monitor='val_accuracy', patience = 5, restore_best_weights = True, verbose = 1),
    ReduceLROnPlateau(monitor='val_loss', factor = 0.5, patience = 3, verbose = 1, min_lr = 1e-7)
]

# Train model
print("Training CNN model")

start_time_cnn = time.time()

history_cnn = cnn_model.fit(
    X_train_padded,
    y_train_binary,
    batch_size = 128,
    epochs = 20,
    validation_split = 0.1,
    callbacks = callbacks_cnn,
    verbose = 1
)

training_time_cnn = time.time() - start_time_cnn

print(f"\nTraining complete in {training_time_cnn:.2f} seconds ({training_time_cnn/60:.1f} minutes)")
print(f"Epochs completed: {len(history_cnn.history['loss'])}")
```

# III. Conclusions

## 9. Performance Analysis & Comparative Discussion

This project implemented and compared two distinct approaches to sentiment analysis on the IMDB movie review dataset: a statistical model (Naive Bayes with TF-IDF features) and an embedding-based model (CNN with pre-trained GloVe embeddings). The comparative analysis reveals important insights about the trade-offs between traditional and modern NLP techniques.

#### Performance:

The embedding-based CNN model achieved superior performance with 87.99% test accuracy, outperforming the statistical Naive Bayes model's 86.39% by 1.6 percentage points. Both models demonstrated balanced performance across positive and negative sentiment classes with CNN showing slight improvements in both precision and recall. The CNN model also show significantly better confidence calibration which give more reliable probability estimates.

#### Visualisation Insights:

The training history plots reveal CNN's learning dynamics, with validation accuracy peaking at epoch 6 with 88.48%, while training accuracy continued 
rising to 96.51% by epoch 11, indicating some overfitting This suggests that despite dropout regularization, the model memorized some training-specific patterns. The class-wise performance comparison shows both models handle positive and negative reviews with similar effectiveness, confirming balanced dataset learning without significant class bias.

#### Critical Evaluation:

The statistical approach's primary advantages lie in computational efficiency and interpretability. Naive Bayes trained in 0.20 seconds which is more than 200 time over CNN, making it ideal for real-time applications and resource-constrained environments. Its probabilistic framework provides transparent word-level importance scores, enabling straightforward model debugging and feature analysis. However, the independence assumption inherent in Naive Bayes fails to capture contextual relationships.

Meanwhile the embedding-based CNN leverages pre-trained semantic knowledge from 6 billion tokens. THis enable it to understand semantic meaning between the words. Its convolutional filters capture n-gram patterns (3-5 word phrases). This semantic understanding and pattern recognition
capability explains the 1.6% accuracy improvement. However, these benefits come at significant computational cost and reduced interpretability.

#### Choice Preference Based on Scenario:

For production deployment requiring real-time predictions with limited computational resources, Naive Bayes is preferable despite lower accuracy. Conversely, applications prioritizing accuracy over speed—such as batch processing of large review datasets or research contexts benefit from CNN's
superior performance. The embedding-based approach particularly excels when semantic understanding matters such as detecting subtle sentiment nuances or handling domain-specific vocabulary through transfer learning.

## 10. Project summary and Reflection

This project provided valuable insights into the practical considerations of choosing between traditional and modern NLP approaches. A key learning was that sophisticated models do not always guarantee superior performance. The Naive Bayes model achieved 86.39% accuracy with minimal
computational overhead, while the CNN required 2,000 times more training time for only 1.6% improvement. This reinforces the principle that effective feature engineering and task-appropriate model selection often outweigh architectural complexity. The hands-on experience with both statistical and deep learning approaches revealed the importance of understanding trade-offs such as interpretability versus accuracy, speed versus performance, and simplicity versus capability.

The statistical model demonstrated exceptional practicality for production deployment. Its 0.20 second training time, zero external dependencies and interpretable probability scores make it ideal for real-world applications requiring rapid deployment, easy debugging and real-time predictions. The
embedding-based CNN model despite achieving higher accuracy requires pre-trained embeddings  and substantial computational resources, making it more suitable for offline batch processing or research environments where accuracy justifies increased complexity.

This work contributes empirical evidence that for well-defined sentiment analysis tasks with clear discriminative vocabulary, traditional statistical models remain better fit. The comprehensive comparison framework provides a reusable methodology for evaluating NLP models. The finding that pre-trained embeddings and n-gram pattern recognition offer marginal gains on IMDB reviews suggests that future sentiment analysis research should focus on more challenging datasets where semantic understanding provides clearer advantages.

The comparative analysis framework transfers readily to similar binary text classification tasks: spam detection, fake news identification or document categorization. The methodology of implementing both statistical baselines and embedding-based models, then evaluating across multiple dimensions (accuracy, speed, interpretability) provides a base for domain-specific applications.

Several enhancements could strengthen this work. First, implementing ensemble methods combining Naive Bayes and CNN predictions might capture benefits of both approaches. Second, exploring modern transformer architectures (BERT, RoBERTa) would likely achieve 92-95% accuracy. Third, investigating active learning could reduce labeling requirements while maintaining accuracy. Fourth, applying explainability techniques to the CNN model would address the interpretability gap with statistical models. Finally, evaluating on domain-specific datasets would test generalization beyond movie reviews.

Future research should explore hybrid architectures that combine interpretable statistical features with semantic embeddings that could potentially achieve both accuracy and explainability. Additionally, investigating few-shot learning approaches could enable rapid adaptation to new sentiment analysis domains with minimal labeled data and address a key practical limitation of deep learning models in resource-constrained scenarios.
